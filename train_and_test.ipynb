```python
import class_ML_carbon as carb
import numpy as np
from netCDF4 import Dataset
```

# read in your data in a netCDF format. The data are in the form of datapoints x features. they are stored in two arrays: "inputs" and "outputs". This file has been saved for separate years and compressed due to GitHub requirements, it would need to be rebuild.

```python
inpfile = Dataset("./data/Free_run_2016-2020_data.nc")   
```

```python
inputs = inpfile.variables["inputs"][:].transpose()   
outputs = inpfile.variables["outputs"][:].transpose()
```

# select the training and validation subset from the data

```python
inputs_train_val = inputs[:round(0.8*inputs.shape[0]),:]
outputs_train_val = outputs[:round(0.8*inputs.shape[0]),:]
```

# select the test subset from the data

```python
inputs_test = inputs[round(0.8*inputs.shape[0]):,:]
outputs_test = outputs[round(0.8*inputs.shape[0]):,:]
```

# use predefined model architecture and train the model on the selected data, after that save the model

```python
model_init = carb.ML_carbon(inputs = inputs_train_val, outputs = outputs_train_val, normalization_inputs = inputs_train_val, normalization_outputs = outputs_train_val, save_model = "./model")
model = model_init.trained_model()
```

# after you developed the model use it to predict the test data 

```python
pred_init = carb.ML_carbon(inputs = inputs_test, model = model)
pred = pred_init.predicted_values()
```

# plot SHAP to explain the model features

```python
shap_init = carb.ML_carbon(inputs = inputs_train_val, model = model)
pred = shap_init.shap_plot()
```
